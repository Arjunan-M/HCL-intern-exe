```python
blurred_image = cv2.GaussianBlur(gray_image,(5, 5),0)
mean_filtered = cv2.blur( gray_image,(5, 5))
gaussian_filtered = cv2.GaussianBlur(gray_image,(5, 5),0)

plt.figure(figsize=(15, 8))

plt.subplot(2, 2, 1)
plt.imshow(rgb_image)
plt.title("Original Industrial Image")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(gray_image, cmap="gray")
plt.title("Grayscale Image")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(mean_filtered, cmap="gray")
plt.title("Mean Filter (5x5)")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(gaussian_filtered, cmap="gray")
plt.title("Gaussian Filter (5x5)")
plt.axis("off")

plt.tight_layout()
plt.show()

```

```python
sobel_x=cv2.Sobel(blurred_image,cv2.CV_64F,1,0,ksize=3)
sobel_y=cv2.Sobel(blurred_image,cv2.CV_64F,0,1,ksize=3)
sobel_magnitude=np.sqrt(sobel_x**2+sobel_y**2)
sobel_magnitude=cv2.convertScaleAbs(sobel_magnitude)

canny_edges=cv2.Canny(blurred_image,100,200)

laplacian=cv2.Laplacian(blurred_image,cv2.CV_64F)
laplacian=cv2.convertScaleAbs(laplacian)

plt.figure(figsize=(15, 10))

plt.subplot(2, 3, 1)
plt.imshow(rgb_image)
plt.title("Original RGB Image")
plt.axis("off")

plt.subplot(2, 3, 2)
plt.imshow(gray_image, cmap="gray")
plt.title("Grayscale Image")
plt.axis("off")

plt.subplot(2, 3, 3)
plt.imshow(blurred_image, cmap="gray")
plt.title("Gaussian Smoothed")
plt.axis("off")

plt.subplot(2, 3, 4)
plt.imshow(sobel_magnitude, cmap="gray")
plt.title("Sobel Edge Detection")
plt.axis("off")

plt.subplot(2, 3, 5)
plt.imshow(canny_edges, cmap="gray")
plt.title("Canny Edge Detection")
plt.axis("off")

plt.subplot(2, 3, 6)
plt.imshow(laplacian, cmap="gray")
plt.title("Laplacian Edge Detection")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_iris
from sklearn import *

iris=load_iris()

df=pd.DataFrame(iris.data,columns=iris.feature_names)

df["species"]=iris.target_names[iris.target]

print("Original Iris Dataset:")
print(df.head())

print("TASK 1: MISSING VALUE HANDLING")

missing_df = df.copy()

missing_df.loc[5, "sepal length (cm)"] = np.nan
missing_df.loc[10, "sepal width (cm)"] = np.nan
missing_df.loc[20, "petal length (cm)"] = np.nan
missing_df.loc[25, "petal width (cm)"] = np.nan
missing_df.loc[30, "species"] = np.nan

print("\nDataset with Missing Values:")
print(missing_df.head(35))

print("\nNumber of Missing Values:")
print(missing_df.isnull().sum())


mean_df = missing_df.copy()

for column in iris.feature_names:
    mean_df[column] = mean_df[column].fillna(
        mean_df[column].mean()
    )
mean_df["species"] = mean_df["species"].fillna(
    mean_df["species"].mode()[0]
)
median_df = missing_df.copy()
for column in iris.feature_names:
    median_df[column] = median_df[column].fillna(
        median_df[column].median()
    )
median_df["species"] = median_df["species"].fillna(
    median_df["species"].mode()[0]
)
mode_df = missing_df.copy()

for column in iris.feature_names:
    mode_df[column] = mode_df[column].fillna(
        mode_df[column].mode()[0]
    )
mode_df["species"] = mode_df["species"].fillna(
    mode_df["species"].mode()[0]
)

print("\nOriginal value at row 5:")
print(missing_df.loc[5, "sepal length (cm)"])

print("\nMean Imputation:")
print(mean_df.loc[5, "sepal length (cm)"])

print("\nMedian Imputation:")
print(median_df.loc[5, "sepal length (cm)"])

print("\nMode Imputation:")
print(mode_df.loc[5, "sepal length (cm)"])
```

```python
print("TASK 2: OUTLIER DETECTION USING IQR")

outlier_df = df.copy()

def find_outliers_iqr(data, column):
    Q1=data[column].quantile(0.25)
    Q3=data[column].quantile(0.75)

    IQR=Q3-Q1

    lower=Q1-1.5*IQR
    upper=Q3+1.5*IQR
    outliers=data[(data[column] < lower) | (data[column] > upper)]
    
    return Q1, Q3, IQR, lower, upper, outliers

for column in iris.feature_names:
    Q1,Q3,iqr,lower,upper,outliers=find_outliers_iqr(outlier_df, column)
    print("\nFeature:", column)
    print("Q1:", Q1)
    print("Q3:", Q3)
    print("IQR:", iqr)
    print("Lower Limit:", lower)
    print("Upper Limit:", upper)
    print("Number of Outliers:", len(outliers))


plt.figure(figsize=(12, 6))
sns.boxplot(data=outlier_df[iris.feature_names])
plt.title("Iris Dataset - Before Outlier Removal")
plt.xticks(rotation=20)
plt.show()
```

```python
clean_df = outlier_df.copy()
for column in iris.feature_names:
    Q1=clean_df[column].quantile(0.25)
    Q3=clean_df[column].quantile(0.75)
    IQR=Q3-Q1
    lower=Q1-1.5*IQR
    upper=Q3+1.5*IQR
    clean_df=clean_df[(clean_df[column] >= lower) & (clean_df[column] <= upper)]
plt.figure(figsize=(12, 6))
sns.boxplot(data=clean_df[iris.feature_names])
plt.title("Iris Dataset - After Outlier Removal")
plt.xticks(rotation=20)
plt.show()
```

```python
print("TASK 3: FEATURE SCALING")
X = df[iris.feature_names]
minmax_scaler=MinMaxScaler()
X_minmax=minmax_scaler.fit_transform(X)
minmax_df=pd.DataFrame(X_minmax,columns=iris.feature_names)

standard_scaler=StandardScaler()
X_standard=standard_scaler.fit_transform(X)
standard_df = pd.DataFrame(X_standard,columns=iris.feature_names)
print("\nOriginal Data:")
print(X.head())
print("\nMin-Max Scaled Data:")
print(minmax_df.head())
print("\nStandardized Data:")
print(standard_df.head())
```

```python
print("TASK 4: CATEGORICAL ENCODING")
encoding_df = df.copy()
label_encoder = LabelEncoder()
encoding_df["species_label"]=label_encoder.fit_transform(encoding_df["species"])
print("\nLabel Encoding:")
print(encoding_df[["species", "species_label"]].head(10))

onehot_encoder=OneHotEncoder(sparse_output=False)
onehot_values = onehot_encoder.fit_transform(encoding_df[["species"]])
onehot_columns = onehot_encoder.get_feature_names_out(["species"])
onehot_df = pd.DataFrame(onehot_values,columns=onehot_columns)

print("\nOne-Hot Encoding:")
print(onehot_df.head(10))
```

```python
print("TASK 5: COMPLETE PREPROCESSING PIPELINE")
pipeline_df = df.copy()
pipeline_df.loc[5, "sepal length (cm)"]=np.nan
pipeline_df.loc[10, "petal width (cm)"]=np.nan
pipeline_df.loc[20, "species"]=np.nan

print("\nDataset Before Preprocessing:")
print(pipeline_df.head(25))
print("\nMissing Values:")
print(pipeline_df.isnull().sum())

numerical_features = iris.feature_names
categorical_features = ["species"]

for column in numerical_features:

    pipeline_df[column] = pipeline_df[column].fillna(
        pipeline_df[column].median()
    )

for column in numerical_features:

    Q1 = pipeline_df[column].quantile(0.25)
    Q3 = pipeline_df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    pipeline_df = pipeline_df[
        (pipeline_df[column] >= lower) &
        (pipeline_df[column] <= upper)
    ]

numeric_pipeline = Pipeline([("imputer",SimpleImputer(strategy="median")),("scaler",StandardScaler())])
categorical_pipeline = Pipeline([("imputer",SimpleImputer(strategy="most_frequent")),( "encoder", OneHotEncoder( handle_unknown="ignore",sparse_output=False))])

preprocessor = ColumnTransformer([ ("numerical",numeric_pipeline,numerical_features),("categorical",categorical_pipeline,categorical_features)])

processed_data = preprocessor.fit_transform(
    pipeline_df
)

encoder = (preprocessor.named_transformers_["categorical"].named_steps["encoder"])
encoded_columns = encoder.get_feature_names_out(categorical_features)
final_columns = (
    numerical_features +
    list(encoded_columns)
)
final_df = pd.DataFrame(
    processed_data,
    columns=final_columns
)

print("FINAL CLEANED DATASET")

print(final_df.head(20))

print("\nMissing Values in Final Dataset:")
print(final_df.isnull().sum())

print("\nFinal Dataset Shape:")
print(final_df.shape)
```
